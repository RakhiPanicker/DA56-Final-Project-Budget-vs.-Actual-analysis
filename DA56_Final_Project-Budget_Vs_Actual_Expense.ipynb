### Introduction to Budget vs. Actual Expense Analysis

A **budget** is a financial plan that estimates the amount of money an organization expects to spend. **Actual expenses** represent the amount of money that was actually spent. Comparing budgeted and actual expenses helps identify spending differences and understand how financial resources are utilized.

This project uses Python libraries such as **Pandas, NumPy, Matplotlib, Seaborn, and Plotly** to explore and analyze the City of Providence expense budget dataset. Pandas and NumPy support data manipulation and numerical calculations, while Matplotlib and Seaborn are used for static visualizations. Plotly enables the creation of interactive charts to present meaningful insights into budget and actual expense patterns.

### Project Purpose and Scope

Finance is a part of everyone's life. Whether managing a household budget or allocating funds across a large organization, the challenge is similar: spending within available resources, prioritizing essential needs, preparing for unexpected expenses, and making informed financial decisions.

Through this project, I aim to use data analytics to understand how the City of Providence allocates and spends its budget, identify departments and expense categories where actual expenses exceed approved amounts, and explore practical ways to improve budget planning, expenditure control, and resource allocation.

The analysis focuses on fiscal years **2012 to 2019**, comparing recommended, approved, and actual expenses to identify spending patterns and meaningful financial insights. The findings will be used to suggest evidence-based improvements in budget planning and spending management.

#### 3. Standardizing Column Names and Converting Financial Data Types

In this step, extra spaces are removed from column names, and the financial amount columns are cleaned and converted to numeric data types. This preparation allows accurate calculations for budget variance and budget utilization.

```python
# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Convert financial amount columns to numeric
amount_columns = ["Recommended Amount", "Approved Amount", "Actual Amount"]

for col in amount_columns:
    df[col] = pd.to_numeric(
        df[col].str.replace(",", "", regex=False),
        errors="coerce"
    )

# Check the updated data types
print(df[amount_columns].dtypes)
```

Info: The extra spaces in the column names were removed, and the Recommended Amount, Approved Amount, and Actual Amount columns were converted from strings to numeric data types (`float64`). Commas were removed from the financial values before conversion, allowing the columns to be used for mathematical calculations and further budget analysis.

```python
# Check for missing values after numeric conversion
print(df[amount_columns].isnull().sum())
```

Info: After converting the financial amount columns to numeric data types, the columns were checked again for missing values. The output showed zero missing values in Recommended Amount, Approved Amount, and Actual Amount. This confirms that the conversion did not introduce any missing values in these columns.

#### 4.Creating the Budget Variance Column

The Budget Variance column is created by subtracting the Approved Amount from the Actual Amount. This helps identify whether actual spending was above, below, or equal to the approved budget.

```python
# Create Budget Variance column
df["Budget Variance"] = df["Actual Amount"] - df["Approved Amount"]

# Display the first five rows
print(df[["Approved Amount", "Actual Amount", "Budget Variance"]].head())
```

Info: The Budget Variance column was created by subtracting the Approved Amount from the Actual Amount. The output shows the difference between actual spending and the approved budget for the first five records. Since the Approved Amount is zero in these records, the Budget Variance equals the Actual Amount. Positive values indicate spending above the approved budget, while negative values indicate spending below the approved budget.

#### 5. Creating the Budget Utilization Percentage Column

The Budget Utilization Percentage column is created to measure the percentage of the approved budget spent based on the actual expenses. This helps compare spending levels against approved budgets. Records with a zero approved amount will need special handling because division by zero is undefined.

```python
# Calculate Budget Utilization Percentage
# Handle zero approved amounts safely

df["Budget Utilization Percentage"] = (
    df["Actual Amount"] / df["Approved Amount"].replace(0, float("nan"))
) * 100

# Display the result
df[["Approved Amount", "Actual Amount", "Budget Utilization Percentage"]].head()
```

**Info Note: Budget Utilization Percentage**: 
The Budget Utilization Percentage column measures the percentage of the approved budget spent based on actual expenses. It is calculated by dividing the Actual Amount by the Approved Amount and multiplying the result by 100. When the Approved Amount is zero, the utilization percentage is recorded as `NaN` because division by zero is undefined. These values are retained rather than replaced with zero to preserve the meaning of the financial data.

### 1.3 Year-wise Budget Utilization Analysis

This visualization examines the percentage of approved budgets utilized through actual expenses during 2012–2019. A line chart is used to compare annual budget utilization rates and observe changes in spending relative to approved budgets over the analysis period.

```python
# Calculate annual budget utilization percentage
yearly_utilization = (
    df_analysis.groupby("Year")
    .agg(
        Total_Approved=("Approved Amount", "sum"),
        Total_Actual=("Actual Amount", "sum")
    )
)

# Calculate utilization only where the approved budget is non-zero
yearly_utilization["Budget Utilization (%)"] = (
    yearly_utilization["Total_Actual"]
    .div(yearly_utilization["Total_Approved"].replace(0, float("nan")))
    * 100
)

# Plot yearly budget utilization
yearly_utilization["Budget Utilization (%)"].plot(
    kind="line", marker="o", figsize=(10, 6)
)

plt.title("Year-wise Budget Utilization Percentage (2012–2019)")
plt.xlabel("Year")
plt.ylabel("Budget Utilization (%)")
plt.xticks(yearly_utilization.index)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

**Info Note:** This line chart illustrates the annual budget utilization percentage for the period 2012–2019.
